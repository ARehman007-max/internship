```python
import pandas as pd
import numpy as np
import io
from google.colab import files
uploaded=files.upload()
```

```python
df=pd.read_csv(io.BytesIO(uploaded['titanic.csv']))
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# 1. Matplotlib histogram
plt.figure(figsize=(6, 4))
plt.hist(df['Age'].dropna(), bins=20, color='skyblue')
plt.title('Distribution of Passenger Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# 2. Seaborn boxplot
plt.figure(figsize=(6, 4))
sns.boxplot(x='Pclass', y='Fare', data=df)
plt.title('Fare Distribution Across Passenger Classes')
plt.show()

# 3. Correlation heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(['Did Not Survive', 'Survived'])
plt.show()
```

```python
survival_counts = df['Survived'].value_counts()
plt.figure(figsize=(6, 4))
plt.pie(survival_counts, labels=['Did Not Survive', 'Survived'],
        autopct='%1.1f%%', colors=['lightcoral', 'lightgreen'], startangle=90)
plt.title('Survival Rate')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.violinplot(x='Pclass', y='Age', data=df, palette='Set2')
plt.title('Age Distribution by Passenger Class')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(df['Age'], df['Fare'], alpha=0.5, color='purple')
plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='Embarked', hue='Survived', data=df)
plt.title('Survival by Embarkation Port')
plt.xlabel('Port (C= Cherbourg, Q= Queenstown, S= Southampton)')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.hist(df['Fare'], bins=30, color='orange', edgecolor='black')
plt.title('Distribution of Fare')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.pairplot(df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].dropna(),
             hue='Survived', palette='Set1')
plt.show()
```
