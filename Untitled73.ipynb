```python
# LLM Hands-on Lab

# %% [markdown]
# # LLM Hands-on Lab

# %%
!pip install transformers torch -q

# %%
from transformers import AutoTokenizer, AutoModelForCausalLM, AutoModel
import torch
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("distilgpt2")

# %% [markdown]
# ## Tokenization Analysis

# %%
texts = [
    "Machine learning models process tokens, not words.",
    "The quick brown fox jumps over the lazy dog.",
    "Transformers are amazing!",
    "Supercalifragilisticexpialidocious",
    "Hello world",
    "Artificial intelligence is transforming society"
]

data = []
for text in texts:
    tokens = tokenizer.encode(text)
    data.append({
        'Text': text[:30] + '...' if len(text) > 30 else text,
        'Characters': len(text),
        'Tokens': len(tokens),
        'Chars_per_Token': len(text)/len(tokens)
    })

df = pd.DataFrame(data)
df['Characters'] = pd.to_numeric(df['Characters'])
df['Tokens'] = pd.to_numeric(df['Tokens'])

display(df)

# %%
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].bar(df['Text'], df['Characters'], alpha=0.7, label='Characters')
axes[0].bar(df['Text'], df['Tokens'], alpha=0.7, label='Tokens')
axes[0].set_xticks(range(len(df['Text'])))
axes[0].set_xticklabels(df['Text'], rotation=45, ha='right')
axes[0].legend()
axes[0].set_title('Characters vs Tokens')

axes[1].scatter(df['Characters'], df['Tokens'], s=100, alpha=0.6)
z = np.polyfit(df['Characters'].values.astype(float), df['Tokens'].values.astype(float), 1)
p = np.poly1d(z)
axes[1].plot(df['Characters'], p(df['Characters']), "r--", alpha=0.8)
axes[1].set_xlabel('Characters')
axes[1].set_ylabel('Tokens')
axes[1].set_title('Correlation')

axes[2].hist(df['Chars_per_Token'], bins=10, edgecolor='black')
axes[2].axvline(4, color='r', linestyle='--', label='Rule of thumb: 4')
axes[2].set_xlabel('Chars per Token')
axes[2].set_ylabel('Frequency')
axes[2].legend()
axes[2].set_title('Chars/Token Distribution')

plt.tight_layout()
plt.show()

# %%
word_counts = np.arange(10, 101, 10)
token_ratios = []
char_ratios = []

for wc in word_counts:
    text = "word " * wc
    tokens = tokenizer.encode(text)
    token_ratios.append(len(tokens)/wc)
    char_ratios.append(len(text)/len(tokens))

fig, ax1 = plt.subplots(figsize=(10, 5))
ax2 = ax1.twinx()
ax1.plot(word_counts, token_ratios, 'b-o', label='Tokens/Word')
ax2.plot(word_counts, char_ratios, 'r-s', label='Chars/Token')
ax1.set_xlabel('Number of Words')
ax1.set_ylabel('Tokens per Word', color='b')
ax2.set_ylabel('Chars per Token', color='r')
ax1.set_title('Token Scaling with Text Length')
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')
plt.show()

# %% [markdown]
# ## Temperature Matrix

# %%
def generate_text(prompt, max_new_tokens=20, do_sample=False, temperature=1.0, top_p=1.0, seed=None):
    if seed:
        torch.manual_seed(seed)
    input_ids = tokenizer.encode(prompt, return_tensors="pt")
    output = model.generate(
        input_ids,
        max_new_tokens=max_new_tokens,
        do_sample=do_sample,
        temperature=temperature,
        top_p=top_p,
        pad_token_id=tokenizer.eos_token_id
    )
    return tokenizer.decode(output[0], skip_special_tokens=True)

# %%
prompt = "The future of AI is"
temperatures = [0.1, 0.5, 0.8, 1.0, 1.5, 2.0]
seeds = [42, 123, 456]

matrix = np.zeros((len(temperatures), len(seeds)), dtype=object)

for i, temp in enumerate(temperatures):
    for j, seed in enumerate(seeds):
        matrix[i][j] = generate_text(prompt, max_new_tokens=10,
                                     do_sample=True, temperature=temp, seed=seed)[len(prompt):]

fig, ax = plt.subplots(figsize=(12, 8))
im = ax.imshow(np.random.rand(len(temperatures), len(seeds)), cmap='RdYlBu_r', aspect='auto')

for i in range(len(temperatures)):
    for j in range(len(seeds)):
        ax.text(j, i, matrix[i][j], ha='center', va='center', fontsize=8, wrap=True)

ax.set_xticks(range(len(seeds)))
ax.set_yticks(range(len(temperatures)))
ax.set_xticklabels([f'Seed {s}' for s in seeds])
ax.set_yticklabels([f'Temp {t}' for t in temperatures])
ax.set_xlabel('Random Seed')
ax.set_ylabel('Temperature')
ax.set_title('Temperature × Seed Generation Matrix')
plt.colorbar(im, ax=ax, label='Diversity')
plt.tight_layout()
plt.show()

# %%
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()

for idx, temp in enumerate(temperatures):
    outputs = []
    for seed in range(5):
        torch.manual_seed(seed)
        input_ids = tokenizer.encode(prompt, return_tensors="pt")
        output = model.generate(input_ids, max_new_tokens=5,
                               do_sample=True, temperature=temp,
                               pad_token_id=tokenizer.eos_token_id)
        outputs.append(tokenizer.decode(output[0], skip_special_tokens=True)[len(prompt):])

    unique_outputs = len(set(outputs))
    axes[idx].bar(range(5), [1]*5, color=plt.cm.viridis(unique_outputs/5))
    axes[idx].set_title(f'Temp={temp}, Unique={unique_outputs}/5')
    axes[idx].set_xticks(range(5))
    axes[idx].set_xticklabels(outputs, fontsize=8)
    axes[idx].set_ylim(0, 1.5)

plt.suptitle('Output Diversity vs Temperature', fontsize=16)
plt.tight_layout()
plt.show()

# %% [markdown]
# ## Top-p Analysis

# %%
top_p_values = [0.1, 0.3, 0.5, 0.7, 0.9, 1.0]
temp_values = [0.3, 0.7, 1.0]

results = {}
for temp in temp_values:
    for top_p in top_p_values:
        outputs = []
        for seed in range(10):
            torch.manual_seed(seed)
            input_ids = tokenizer.encode(prompt, return_tensors="pt")
            output = model.generate(input_ids, max_new_tokens=3,
                                   do_sample=True, temperature=temp, top_p=top_p,
                                   pad_token_id=tokenizer.eos_token_id)
            outputs.append(tokenizer.decode(output[0], skip_special_tokens=True)[len(prompt):])
        results[(temp, top_p)] = len(set(outputs)) / 10

matrix = np.zeros((len(temp_values), len(top_p_values)))
for i, temp in enumerate(temp_values):
    for j, top_p in enumerate(top_p_values):
        matrix[i][j] = results[(temp, top_p)]

fig, ax = plt.subplots(figsize=(10, 6))
im = ax.imshow(matrix, cmap='YlOrRd', aspect='auto')

for i in range(len(temp_values)):
    for j in range(len(top_p_values)):
        ax.text(j, i, f'{matrix[i][j]:.2f}', ha='center', va='center', fontsize=12)

ax.set_xticks(range(len(top_p_values)))
ax.set_yticks(range(len(temp_values)))
ax.set_xticklabels(top_p_values)
ax.set_yticklabels(temp_values)
ax.set_xlabel('Top-p')
ax.set_ylabel('Temperature')
ax.set_title('Diversity Heatmap (Temperature × Top-p)')
plt.colorbar(im, label='Unique Outputs Ratio')
plt.show()

# %% [markdown]
# ## Context Window Visualization

# %%
context_limit = tokenizer.model_max_length
text_lengths = [100, 500, 1000, 1500, 2000, 3000]

fig, ax = plt.subplots(figsize=(12, 4))

for i, length in enumerate(text_lengths):
    color = 'green' if length <= context_limit else 'red'
    ax.barh(i, length, color=color, alpha=0.7)
    ax.text(length + 20, i, f'{length} tokens', va='center')
    if length > context_limit:
        ax.text(length + 20, i + 0.2, 'Exceeds', va='center', color='red', fontsize=8)

ax.axvline(context_limit, color='blue', linestyle='--', linewidth=2, label=f'Context Window: {context_limit}')
ax.set_yticks(range(len(text_lengths)))
ax.set_yticklabels([f'Text {i+1}' for i in range(len(text_lengths))])
ax.set_xlabel('Number of Tokens')
ax.set_title('Context Window Limits')
ax.legend()
plt.tight_layout()
plt.show()

# %%
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].pie([context_limit, 4000 - context_limit],
            labels=[f'GPT-2 ({context_limit})', 'Modern Models (~4000+)'],
            autopct='%1.1f%%', startangle=90, colors=['skyblue', 'lightcoral'])
axes[0].set_title('Context Window Comparison')

models = ['GPT-2', 'GPT-3', 'GPT-3.5', 'GPT-4', 'Claude-3']
windows = [1024, 2048, 4096, 8192, 200000]
axes[1].bar(models, windows, color=plt.cm.plasma(np.linspace(0.2, 0.8, len(models))))
axes[1].set_yscale('log')
axes[1].set_ylabel('Context Window (tokens, log scale)')
axes[1].set_title('Context Window Evolution')
for i, v in enumerate(windows):
    axes[1].text(i, v * 1.1, str(v), ha='center')

plt.tight_layout()
plt.show()

# %% [markdown]
# ## Loss Landscape

# %%
prompt = "The cat"
continuations = [" sat", " ran", " ate", " slept", " jumped"]
temperatures = np.linspace(0.1, 2.0, 10)

fig, ax = plt.subplots(figsize=(10, 6))

for cont in continuations:
    probs = []
    for temp in temperatures:
        torch.manual_seed(42)
        input_text = prompt + cont
        input_ids = tokenizer.encode(input_text, return_tensors="pt")

        with torch.no_grad():
            outputs = model(input_ids)
            logits = outputs.logits[0, -1, :]
            probs.append(torch.softmax(logits / temp, dim=-1).max().item())

    ax.plot(temperatures, probs, marker='o', label=f'"{cont}"')

ax.set_xlabel('Temperature')
ax.set_ylabel('Max Token Probability')
ax.set_title('Effect of Temperature on Token Probabilities')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()

# %%
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

test_prompts = [
    "The capital of France is",
    "2 + 2 =",
    "The color of the sky is",
    "Python is a"
]

for idx, test_prompt in enumerate(test_prompts):
    ax = axes[idx // 2][idx % 2]

    input_ids = tokenizer.encode(test_prompt, return_tensors="pt")
    with torch.no_grad():
        outputs = model(input_ids)
        logits = outputs.logits[0, -1, :]
        probs = torch.softmax(logits, dim=-1)

    top_k = 10
    top_probs, top_indices = torch.topk(probs, top_k)
    top_tokens = [tokenizer.decode([idx]) for idx in top_indices]

    ax.bar(top_tokens, top_probs.numpy(), alpha=0.7, color=plt.cm.viridis(np.linspace(0.2, 0.9, top_k)))
    ax.set_title(f'Next Token Distribution: "{test_prompt}"')
    ax.set_xlabel('Token')
    ax.set_ylabel('Probability')
    ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# %% [markdown]
# ## Attention Pattern

# %%
base_model = AutoModel.from_pretrained("distilgpt2", output_attentions=True)

text = "The cat sat on the mat because it was tired"
input_ids = tokenizer.encode(text, return_tensors="pt")

with torch.no_grad():
    outputs = base_model(input_ids)
    attentions = outputs.attentions

    if attentions and len(attentions) > 0:
        last_layer_att = attentions[-1]
        avg_attention = last_layer_att[0].mean(dim=0).numpy()
        tokens = tokenizer.convert_ids_to_tokens(input_ids[0])

        fig, ax = plt.subplots(figsize=(10, 8))
        im = ax.imshow(avg_attention, cmap='Blues', aspect='auto')

        ax.set_xticks(range(len(tokens)))
        ax.set_yticks(range(len(tokens)))
        ax.set_xticklabels(tokens, rotation=45, ha='right')
        ax.set_yticklabels(tokens)
        ax.set_xlabel('Key Tokens')
        ax.set_ylabel('Query Tokens')
        ax.set_title('Average Attention Pattern (Last Layer)')
        plt.colorbar(im, label='Attention Weight')
        plt.tight_layout()
        plt.show()
    else:
        print("No attention weights available.")

# %%
if attentions and len(attentions) >= 6:
    fig, axes = plt.subplots(1, 4, figsize=(16, 4))
    layers_to_plot = [0, 2, 4, 5]
    for i, layer_idx in enumerate(layers_to_plot):
        attention = attentions[layer_idx][0].mean(dim=0).numpy()

        ax = axes[i]
        im = ax.imshow(attention, cmap='Reds', aspect='auto')
        ax.set_title(f'Layer {layer_idx + 1}')
        ax.set_xticks(range(len(tokens)))
        ax.set_yticks(range(len(tokens)))
        ax.set_xticklabels(tokens, rotation=45, ha='right', fontsize=8)
        ax.set_yticklabels(tokens, fontsize=8)

    plt.suptitle('Attention Patterns Across Layers')
    plt.tight_layout()
    plt.show()

# %% [markdown]
# ## Performance Metrics

# %%
test_cases = [
    ("The sky is", " blue"),
    ("I love to", " eat"),
    ("The weather today is", " nice"),
    ("Machine learning is", " fun")
]

metrics = []
for prompt, expected in test_cases:
    generated = generate_text(prompt, max_new_tokens=5, do_sample=False)
    metrics.append({
        'Prompt': prompt,
        'Generated': generated,
        'Expected': prompt + expected,
        'Match': generated == prompt + expected
    })

metrics_df = pd.DataFrame(metrics)
display(metrics_df)

accuracy = metrics_df['Match'].mean()
print(f'\nAccuracy: {accuracy:.2%}')

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(['Correct', 'Incorrect'], [accuracy, 1-accuracy], color=['green', 'red'], alpha=0.7)
ax.set_ylabel('Percentage')
ax.set_title('Model Performance on Simple Completions')
for i, v in enumerate([accuracy, 1-accuracy]):
    ax.text(i, v/2, f'{v:.1%}', ha='center', fontsize=14)
plt.show()

# %%
latencies = []
token_counts = [10, 20, 50, 100, 200]

for num_tokens in token_counts:
    import time
    start = time.time()
    generate_text("Hello", max_new_tokens=num_tokens, do_sample=False)
    latencies.append(time.time() - start)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(token_counts, latencies, 'o-', linewidth=2, markersize=8)
ax.set_xlabel('Number of Tokens Generated')
ax.set_ylabel('Latency (seconds)')
ax.set_title('Generation Time vs Token Count')
ax.grid(True, alpha=0.3)

z = np.polyfit(token_counts, latencies, 1)
p = np.poly1d(z)
ax.plot(token_counts, p(token_counts), 'r--', alpha=0.5, label=f'Trend: {z[0]:.4f} sec/token')
ax.legend()
plt.show()

# %%
del model, base_model, tokenizer
torch.cuda.empty_cache() if torch.cuda.is_available() else None
print(" Done")
```
