```python
from transformers import AutoModelForCausalLM, pipeline, AutoTokenizer
import torch

# Load GPT-2 tokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")

# Load GPT-2 model
model = AutoModelForCausalLM.from_pretrained("gpt2")

# Create a text generation pipeline
text_generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    device=0 if torch.cuda.is_available() else -1 # Use GPU if available
)

def generate_text(prompt, max_new_tokens, do_sample, temperature=1.0, top_p=1.0):
    """Wrapper for the text generation pipeline."""
    # The pipeline returns a list of dictionaries, we want the generated_text from the first one
    result = text_generator(
        prompt,
        max_new_tokens=max_new_tokens,
        do_sample=do_sample,
        temperature=temperature,
        top_p=top_p,
        return_full_text=False # Only return the generated text
    )
    return result[0]['generated_text']

prompts = [
    "The future of artificial intelligence is",
    "In a world where robots",
    "The key to happiness is",
    "Machine learning will"
]

print(" Basic Generation (Greedy - no sampling):\n")
for prompt in prompts:
    result = generate_text(prompt, max_new_tokens=20, do_sample=False)
    print(f"Prompt: {prompt}")
    print(f"Generated: {result}\n")

# %% [markdown]
# ## Sampling Parameters: Temperature & Top-p
#
# **Temperature** controls randomness:
# - Low (0.2): More deterministic, repetitive
# - High (1.5): More random, creative

# %% [markdown]
print(" Temperature Effects:\n")

prompt = "The cat sat on the"
temperatures = [0.1, 0.5, 1.0, 1.5]

for temp in temperatures:
    print(f"Temperature = {temp}:")
    for i in range(3):  # Generate 3 examples
        result = generate_text(prompt, max_new_tokens=15, do_sample=True, temperature=temp)
        print(f"  {result}")
    print()

# %% [markdown]
# ### Top-p (Nucleus Sampling)

# %% [markdown]
print(" Top-p Effects (with fixed temperature):\n")

top_p_values = [0.3, 0.6, 0.9, 1.0]
prompt = "Scientists have discovered"

for top_p in top_p_values:
    print(f"top_p = {top_p}:")
    result = generate_text(prompt, max_new_tokens=20, do_sample=True, temperature=0.8, top_p=top_p)
    print(f"  {result}")
    print()

# %% [markdown]
# ##  Context Windows

# %% [markdown]
# Check model's context window
max_context = tokenizer.model_max_length
print(f" GPT-2 Context Window: {max_context} tokens")

# Test context window limits
test_texts = [
    "Short text under limit",  # Very short
    "word " * 500,  # 500 tokens
    "word " * 1000,  # ~1000 tokens (near limit)
    "word " * 1500,  # Exceeds limit
]

print("\n Token counts for different text lengths:")
for text in test_texts:
    tokens = tokenizer.encode(text)
    print(f"  Length: {len(text):6d} chars → {len(tokens):6d} tokens (within limit: {len(tokens) <= max_context})")

# %% [markdown]
# ##  Understanding LLM Limitations

# %% [markdown]
print(" Common LLM Limitations:\n")

# Test 1: Hallucination potential
print(" Hallucination Test:")
result = generate_text("The president of Mars in 2025 is", max_new_tokens=20, do_sample=True, temperature=0.7)
print(f"   Generated: {result}")
print("   Note: Mars has no president - this is likely hallucinated!\n")

# Test 2: Arithmetic errors
print(" Arithmetic Test:")
result = generate_text("What is 17 * 24? The answer is", max_new_tokens=10, do_sample=False)
print(f"   Generated: {result}")
print(f"   Actual answer: {17 * 24} (models often struggle with arithmetic)\n")

# Test 3: Knowledge cutoff
print(" Knowledge Cutoff Test:")
result = generate_text("The most recent iPhone model is the iPhone", max_new_tokens=5, do_sample=False)
print(f"   Generated: {result}")
print("   Note: This model's training data ends in 2019, so it doesn't know about newer iPhones!")

# %% [markdown]
# ##  Interactive Demo
# Create your own text generation playground!

# %% [markdown]
from ipywidgets import interact, widgets
from IPython.display import display, HTML

def interactive_generation(prompt, temperature, top_p, max_tokens, num_samples):
    """Interactive text generation with adjustable parameters"""
    results = []
    for _ in range(num_samples):
        result = generate_text(
            prompt,
            max_new_tokens=max_tokens,
            do_sample=True,
            temperature=temperature,
            top_p=top_p
        )
        results.append(result)

    html = "<div style='background-color: #f0f0f0; padding: 15px; border-radius: 5px;'>"
    html += f"<h3>Generated Results:</h3>"
    for i, result in enumerate(results, 1):
        html += f"<p><strong>Sample {i}:</strong> {result}</p>"
    html += "</div>"
    display(HTML(html))

# Create interactive widgets
prompt_input = widgets.Text(
    value="Once upon a time",
    description='Prompt:',
    style={'description_width': 'initial'}
)

temp_slider = widgets.FloatSlider(
    value=0.8, min=0.1, max=2.0, step=0.1,
    description='Temperature:',
    style={'description_width': 'initial'}
)

top_p_slider = widgets.FloatSlider(
    value=0.9, min=0.1, max=1.0, step=0.05,
    description='Top-p:',
    style={'description_width': 'initial'}
)

max_tokens_slider = widgets.IntSlider(
    value=30, min=5, max=100, step=5,
    description='Max tokens:',
    style={'description_width': 'initial'}
)

samples_slider = widgets.IntSlider(
    value=3, min=1, max=5, step=1,
    description='Samples:',
    style={'description_width': 'initial'}
)

# Display interactive interface
print(" Interactive Text Generation Playground:")
print("Adjust the parameters and watch how the output changes!\n")

interact(
    interactive_generation,
    prompt=prompt_input,
    temperature=temp_slider,
    top_p=top_p_slider,
    max_tokens=max_tokens_slider,
    num_samples=samples_slider
)

# %% [markdown]
# ##  Key Takeaways
#
# ### What We Learned:
#
# 1. **Tokenization**: LLMs process tokens (~4 chars or 0.75 words each), not raw text
# 2. **Generation**: Models predict one token at a time based on patterns learned during training
# 3. **Sampling Parameters**:
#    - Temperature controls randomness (higher = more creative, lower = more focused)
#    - Top-p restricts sampling to likely tokens (nucleus sampling)
# 4. **Context Windows**: Models have token limits (GPT-2: 1,024 tokens)
# 5. **Limitations**:
#    - Hallucination: Fluent but incorrect statements
#    - Knowledge cutoff: No awareness of events after training
#    - No true reasoning: Can make basic logical/arithmetic errors
#    - Bias: Reflects biases in training data
#
# ### Pro Tips:
# - Start with temperature=0.7 and top_p=0.9 for balanced output
# - Use lower temperature (0.2-0.5) for factual/code generation
# - Use higher temperature (1.0-1.5) for creative writing
# - Always verify important facts from LLM outputs
# - Be mindful of token limits when working with long texts

# %% [markdown]
# Clean up
import gc
del model, tokenizer
gc.collect()
torch.cuda.empty_cache() if torch.cuda.is_available() else None
print(" Session completed successfully!")
```
